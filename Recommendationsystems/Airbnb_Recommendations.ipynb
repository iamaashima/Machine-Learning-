### Determine Cosine Similarity

````python
# define two lists or arrays to compare
A = np.squeeze(np.asarray(sd_pp_array))
B = np.squeeze(np.asarray(random_listing_array))
print("A:\n", A)
print("B:\n", B)
 
# compute cosine similarity
cosine = np.dot(A,B)/(norm(A, axis=1)*norm(B))
print("Cosine Similarity:\n", cosine)
````

You're using **one random listing as a query** and comparing it with **every other listing** in the dataset. This is exactly how a content-based recommendation system finds similar items.

### Step-by-step

```python
random_listing = sd_pp.sample(n=1)
```

* Randomly selects **1 Airbnb listing** from the dataset.
* This acts as the **reference listing** (the one for which we want recommendations).

---

```python
random_listing_array = random_listing.values
```

* Converts the selected listing (1 row) into a NumPy array.

Example:

```
[[0.82, 1.45, 0, 1, 0.67, ...]]
```

---

```python
sd_pp_array = sd_pp.values
```

* Converts the **entire dataset** into a NumPy array.

Example:

```
[
 [0.82, 1.45, ...],
 [0.23, 0.56, ...],
 [1.12, 0.78, ...],
 ...
]
```

---

```python
A = np.squeeze(np.asarray(sd_pp_array))
```

* `A` contains **all listings**.
* Shape:

```
(number_of_listings, number_of_features)
```

---

```python
B = np.squeeze(np.asarray(random_listing_array))
```

* `B` is the **single selected listing**.
* Shape:

```
(number_of_features,)
```

---

```python
cosine = np.dot(A,B)/(norm(A, axis=1)*norm(B))
```

This calculates the cosine similarity between:

* **Every row in A** (every Airbnb listing)
* **The one listing B** (the randomly selected listing)

The formula is

[
\text{Cosine Similarity}=\frac{A\cdot B}{|A||B|}
]

Since `A` has many rows, NumPy computes the dot product of **each row** with `B`.

The result is something like

```
Listing 1 → 0.92
Listing 2 → 0.34
Listing 3 → 0.81
Listing 4 → 0.15
...
```

So you get **one similarity score for every listing**.

### Why compare with just one row?

Because in a recommendation system, you usually start with **one listing the user likes**.

For example:

```
User likes Listing #250
        ↓
Compare Listing #250 with every other listing
        ↓
Similarity scores
        ↓
Recommend the top 10 most similar listings
```

If you compared **every listing with every other listing**, you'd get an (N \times N) similarity matrix, which is more expensive and only needed if you want similarities between all pairs of listings.

### In short

* **One row (`B`)** = the query or reference listing.
* **All rows (`A`)** = candidate listings.
* **Cosine similarity** measures how similar each listing is to the reference listing.
* Listings with the **highest cosine similarity (closest to 1)** are recommended because they have the most similar features.


Yes, exactly.

This is a **content-based recommendation system** because recommendations are based on the **features (content)** of the listings, not on other users' behavior.

The process is:

1. Select a listing that the user is interested in (the random listing in your code).
2. Represent that listing using its features (price, room type, property type, amenities, neighborhood, review score, etc.).
3. Compare those features with every other listing using **cosine similarity**.
4. Recommend the listings with the highest similarity scores.

For example:

* **User likes Listing A**

  * Entire home
  * Apartment
  * Downtown
  * Price: $120
  * Rating: 4.8

The system finds other listings with similar characteristics:

* ✅ Entire home
* ✅ Apartment
* ✅ Similar neighborhood
* ✅ Similar price
* ✅ Similar rating

These become the recommendations.

### Why is it called "content-based"?

Because it uses the **content (attributes/features) of the items** themselves.

It does **not** use:

* what other users viewed,
* what other users liked,
* purchase history of similar users.

That would be **collaborative filtering**.


> This is a content-based recommendation system because it recommends listings by comparing the features of the listing the user is interested in with the features of all other listings using cosine similarity.

````python
def get_recommendations(df, listing):
    """
    Takes in preprocessed dataframe and selected listing as inputs and gives top 5 (including listing)
    recommendations based on cosine similarity. 
    """
    # reset the index
    df = df.reset_index(drop='index')
    #convert single listing to an array
    listing_array = listing.values
    #convert all listings to an array
    df_array = df.values
    # get arrays into a single dimension
    
    A = np.squeeze(np.asarray(df_array))
    B = np.squeeze(np.asarray(listing_array))
    
    # compute cosine similarity 
    cosine = np.dot(A,B)/(norm(A, axis = 1)*norm(B))
    
    # add similarity into recommendations df and reset the index
    rec = sd_clustered.copy().reset_index(drop = 'index')
    rec['similarity'] = pd.DataFrame(cosine).values
    
    # reorder column names
    rec = rec[['id','listing_url', 'similarity', 'cluster_label', 'latitude', 'longitude',
       'neighbourhood_cleansed', 'zipcode', 'property_type', 'room_type',
       'accommodates', 'bathrooms', 'bedrooms', 'beds', 'bed_type',
       'nightly_price', 'price_per_stay', 'security_deposit', 'cleaning_fee',
       'guests_included', 'extra_people', 'minimum_nights', 'maximum_nights',
       'host_response_time', 'host_response_rate', 'host_is_superhost',
       'host_total_listings_count', 'host_has_profile_pic',
       'host_identity_verified', 'number_of_reviews', 'number_of_stays',
       'review_scores_rating', 'review_scores_accuracy',
       'review_scores_cleanliness', 'review_scores_checkin',
       'review_scores_communication', 'review_scores_location',
       'review_scores_value', 'requires_license', 'instant_bookable',
       'is_business_travel_ready', 'cancellation_policy',
       'require_guest_profile_picture', 'require_guest_phone_verification']]
    #sort by top 5 desceding
    return rec.sort_values(by=['similarity'],ascending = False).head(6)
````

Why is this necessary for your recommendation system?

Your recommendation system is content-based, meaning it compares listings based on their features. Cosine similarity can only compare numerical vectors.

Therefore, before computing similarity:

Categorical features (neighborhood, property type, room type) are converted into binary numeric columns using One-Hot Encoding.
Numeric features (price, beds, bathrooms, ratings, etc.) are standardized so that no single feature dominates because of its larger scale.
ColumnTransformer applies the appropriate preprocessing to each type of feature and combines the results into one numeric feature matrix.

This processed matrix is then used to compute cosine similarity, allowing listings with similar characteristics to be recommended to users.

### Construct a new preprocessing pipeline

````python
#define nominal and ordinal features in the categorical columns
nom_cols = sd_simplified.select_dtypes(['object']).columns
print(nom_cols)
ordinal_cols = sd_simplified.select_dtypes(['category']).columns
print(ordinal_cols)
````

We only have numeric and nominal features to preprocess in this simplified dataframe.

````python
#define numeric transformation pipeline that scales the numbers
numeric_pipeline = Pipeline([('numnorm',StandardScaler())])
# define a nominal transformation pipeline that OHE the cats
nominal_pipeline = Pipeline([('onehotenc', OneHotEncoder(categories= "auto", 
                                                         sparse_output = False, 
                                                         handle_unknown = 'ignore'))]) 
````

````python
# construct column transformer for the selected columns with pipelines
ct = ColumnTransformer(transformers = [("nominalpipe", nominal_pipeline, ['neighbourhood_cleansed', 
                                                                          'property_type', 'room_type']),
                                       ("numericpipe", numeric_pipeline, sd_simplified.select_dtypes(['int', 'float']).columns)])
````


### What are we doing?

We are telling scikit-learn:

* Apply **One-Hot Encoding** (`nominal_pipeline`) to the categorical columns.
* Apply **Standard Scaling** (`numeric_pipeline`) to the numeric columns.
* Combine the transformed columns into a single processed dataset.

### What are `transformers`?

`transformers` is a **list of preprocessing rules**. Each rule has three parts:

```python
(name, transformer, columns)
```

* **name**: A label for the transformation (e.g., `"nominalpipe"`).
* **transformer**: The preprocessing method or pipeline to apply (e.g., `nominal_pipeline`).
* **columns**: The columns on which the transformer should be applied.

In your code:

* `("nominalpipe", nominal_pipeline, [...])` → One-Hot Encodes the categorical columns.
* `("numericpipe", numeric_pipeline, ...)` → Standardizes the numeric columns.


> The `ColumnTransformer` applies different preprocessing techniques to different column types—One-Hot Encoding for categorical features and Standard Scaling for numeric features—and combines the results into a single dataset ready for similarity computation.

````python
def get_simplified_recommendations(df, listing):
    """
    Takes in preprocessed dataframe and preprocessed custom listing as inputs and gives top 5
    recommendations based on cosine similarity. 
    """
    # reset the index
    df = df.reset_index(drop = 'index')
    
    # preprocess the listing input
    listing_pp = pd.DataFrame(ct.transform(listing))
    
    # convert single listing to an array
    listing_array = listing_pp.values
    # convert all listings to an array
    df_array = df.values
    
    # get arrays into a single dimension
    A = np.squeeze(np.asarray(df_array))
    B = np.squeeze(np.asarray(listing_array))
    
    # compute cosine similarity 
    cosine = np.dot(A,B)/(norm(A, axis = 1)*norm(B))
    
    # add similarity into recommendations df and reset the index
    rec = sd_simplified.copy().reset_index(drop = 'index')
    rec['similarity'] = pd.DataFrame(cosine).values
    
    # add in listings_urls
    # merge on index
    rec = rec.join(sd_listings_url)
    
    # reorder column names
    rec = rec[['listing_url', 'similarity', 'neighbourhood_cleansed', 'property_type', 
               'room_type', 'accommodates', 'bathrooms', 'beds', 'nightly_price', 'review_scores_rating']]
    
    # sort by top 5 descending
    return rec.sort_values(by = ['similarity'], ascending = False).head(5)
````

````python
import numpy as np

def precision_at_k(recommended, relevant, k=5):
    recommended = recommended[:k]
    hits = len(set(recommended) & set(relevant))
    return hits / k


def recall_at_k(recommended, relevant, k=5):
    recommended = recommended[:k]
    hits = len(set(recommended) & set(relevant))
    return hits / len(relevant)


def average_precision(recommended, relevant, k=5):
    recommended = recommended[:k]
    relevant = set(relevant)

    score = 0
    hits = 0

    for i, item in enumerate(recommended, start=1):
        if item in relevant:
            hits += 1
            score += hits / i

    if hits == 0:
        return 0

    return score / min(len(relevant), k)
````

````python
K = 5

precision_scores = []
recall_scores = []
ap_scores = []

for idx in sd_clustered.index:

    # Query listing
    query_listing = sd_pp.iloc[[idx]]

    # Get recommendations
    rec = get_recommendations(sd_pp, query_listing)

    # Recommended listing indices
    recommended = rec.index.tolist()

    # Remove the query listing itself if present
    recommended = [i for i in recommended if i != idx]

    # Ground truth = all listings in the same cluster
    cluster = sd_clustered.loc[idx, 'cluster_label']

    relevant = sd_clustered[
        sd_clustered['cluster_label'] == cluster
    ].index.tolist()

    relevant.remove(idx)

    precision_scores.append(
        precision_at_k(recommended, relevant, K)
    )

    recall_scores.append(
        recall_at_k(recommended, relevant, K)
    )

    ap_scores.append(
        average_precision(recommended, relevant, K)
    )

print(f"Precision@{K}: {np.mean(precision_scores):.4f}")
print(f"Recall@{K}: {np.mean(recall_scores):.4f}")
print(f"MAP@{K}: {np.mean(ap_scores):.4f}")
````
